# Run postgis docker container
`docker run --rm -it -p 5432:5432 --name postgis_h3 postgis/postgis:16-master`

# Install the h3 extension
`docker exec -it postgis_h3 bash -c "psql -U postgres -c 'CREATE EXTENSION h3;'"`

# Verify installation
`docker exec -it postgis_h3 bash -c "psql -U postgres -c '\dx;'"`

```python
from sqlalchemy import create_engine
from utils import DB_PARAMS

# Replace with your PostgreSQL credentials
username = DB_PARAMS['username']
password = DB_PARAMS['password']
host = DB_PARAMS['host']  # or your database host
port = DB_PARAMS['port']       # default PostgreSQL port
database = DB_PARAMS['database']
# Create the connection string
connection_string = f'postgresql://{username}:{password}@{host}:{port}/{database}'

# Create the SQLAlchemy engine
engine = create_engine(connection_string)

# Test the connection
try:
    with engine.connect() as connection:
        print("Connection to PostgreSQL database successful!")
except Exception as e:
    print(f"An error occurred: {e}")
```
