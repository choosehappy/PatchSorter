```python
%%time
import pandas as pd
import datashader as ds
import datashader.transfer_functions as tf
from datashader.utils import export_image
```

```python

# load dataset
df = pd.read_parquet("gaussian_overlap_dataset.parquet")

print("Loaded:", df.shape)
```

```python
%%time
# datashader requires categorical
df["label"] = df["label"].astype("category")

# canvas
canvas = ds.Canvas(
    plot_width=1200,
    plot_height=1200,
    x_range=(-1, 1),
    y_range=(-1, 1)
)

# aggregation
agg = canvas.points(
    df,
    x="x",
    y="y",
    agg=ds.count_cat("label")
)
```

```python
%%time
# render
img = tf.shade(
    agg,
    color_key={
        0: "#1f77b4",
        1: "#ff7f0e",
        2: "#2ca02c",
        3: "#d62728",
        4: "#9467bd",
    },
    how="eq_hist"
)

img = tf.set_background(img, "white")
img = tf.dynspread(img, threshold=0.5, max_px=4)
```

```python
%%time
# save image
export_image(img, "datashader_plot")

print("Saved datashader_plot.png")
```
